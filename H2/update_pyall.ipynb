```python
import subprocess, sys

def pip_install(packages, upgrade=False):
    if isinstance(packages, str):
        packages = [packages]

    cmd = [sys.executable, "-m", "pip", "install", "-q"]
    if upgrade:
        cmd.append("-U")
    cmd.extend(packages)

    try:
        subprocess.run(
            cmd,
            check=True,
            stdout=subprocess.DEVNULL,
            stderr=subprocess.PIPE,
            text=True
        )
        print(f"✅ OK: {' '.join(packages)}")
    except subprocess.CalledProcessError as e:
        print(f"❌ ERROR instalando: {' '.join(packages)}")
        print(e.stderr.splitlines()[-5:])


pip_install(["matplotlib"], upgrade=True)
```
